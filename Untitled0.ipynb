```python
import pandas as pd
import os

df = pd.read_csv(os.path.join(path, "US_Regional_Sales_Data.csv"))

df.head()
```

```python
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
    .str.replace("_", "", regex=False)
)

print(df.columns)
```

```python
print("Duplicate rows:", df.duplicated().sum())
```

```python
print("Sales Channels:")
print(df["saleschannel"].value_counts())

print("\nWarehouses:")
print(df["warehousecode"].value_counts())

print("\nCurrency:")
print(df["currencycode"].value_counts())
```

```python
date_columns = [
    "procureddate",
    "orderdate",
    "shipdate",
    "deliverydate"
]

for col in date_columns:
    df[col] = pd.to_datetime(
        df[col],
        dayfirst=True,
        errors="coerce"
    )

df[date_columns].dtypes
```

```python
df[date_columns].isnull().sum()
```

```python
df[date_columns].head()
```

```python
if df["unitcost"].dtype == 'object':
    df["unitcost"] = (
        df["unitcost"]
        .str.replace("$", "", regex=False)
        .str.replace(",", "", regex=False)
        .astype(float)
    )

if df["unitprice"].dtype == 'object':
    df["unitprice"] = (
        df["unitprice"]
        .str.replace("$", "", regex=False)
        .str.replace(",", "", regex=False)
        .astype(float)
    )
```

```python
df[["unitcost", "unitprice"]].dtypes
```

```python
print("Minimum quantity:", df["orderquantity"].min())
print("Maximum quantity:", df["orderquantity"].max())
print("Zero quantities:", (df["orderquantity"] == 0).sum())
print("Negative quantities:", (df["orderquantity"] < 0).sum())
```

```python
print("Minimum unit cost:", df["unitcost"].min())
print("Minimum unit price:", df["unitprice"].min())

print("Zero costs:", (df["unitcost"] == 0).sum())
print("Zero prices:", (df["unitprice"] == 0).sum())
```

```python
print("Price below cost:", (df["unitprice"] < df["unitcost"]).sum())
```

```python
print("Minimum discount:", df["discountapplied"].min())
print("Maximum discount:", df["discountapplied"].max())
```

```python
print(df["discountapplied"].value_counts().sort_index())
```

```python
print(
    "Order before procurement:",
    (df["orderdate"] < df["procureddate"]).sum()
)

print(
    "Ship before order:",
    (df["shipdate"] < df["orderdate"]).sum()
)

print(
    "Delivery before ship:",
    (df["deliverydate"] < df["shipdate"]).sum()
)
```

```python
df["sales"] = (
    df["orderquantity"] *
    df["unitprice"] *
    (1 - df["discountapplied"])
)
```

```python
df["total_cost"] = (
    df["orderquantity"] *
    df["unitcost"]
)
```

```python
df["profit"] = df["sales"] - df["total_cost"]
```

```python
df["profit_margin"] = (
    df["profit"] / df["sales"]
)
```

```python
df["delivery_days"] = (
    df["deliverydate"] - df["shipdate"]
).dt.days
```

```python
df["shipping_days"] = (
    df["shipdate"] - df["orderdate"]
).dt.days
```

```python
df["procurement_days"] = (
    df["orderdate"] - df["procureddate"]
).dt.days
```

```python
df[
    [
        "orderquantity",
        "unitprice",
        "discountapplied",
        "sales",
        "total_cost",
        "profit",
        "profit_margin",
        "delivery_days",
        "shipping_days",
        "procurement_days"
    ]
].head()
```

```python
df[
    [
        "sales",
        "total_cost",
        "profit",
        "profit_margin",
        "delivery_days",
        "shipping_days",
        "procurement_days"
    ]
].describe()
```

```python
loss_orders = df[df["profit"] < 0]

print("Loss-making orders:", len(loss_orders))
print("Percentage of orders:", len(loss_orders) / len(df) * 100)
```

```python
loss_orders[
    [
        "ordernumber",
        "saleschannel",
        "warehousecode",
        "customerid",
        "productid",
        "orderquantity",
        "discountapplied",
        "unitcost",
        "unitprice",
        "sales",
        "total_cost",
        "profit",
        "profit_margin"
    ]
].head(10)
```

```python
loss_orders.groupby("discountapplied").agg(
    orders=("ordernumber", "count"),
    avg_profit=("profit", "mean"),
    total_profit=("profit", "sum")
).sort_index()
```

```python
df["profit_status"] = df["profit"].apply(
    lambda x: "Loss" if x < 0 else "Profit"
)

df.groupby("profit_status").agg(
    orders=("ordernumber", "count"),
    avg_sales=("sales", "mean"),
    avg_profit=("profit", "mean"),
    avg_discount=("discountapplied", "mean")
)
```

```python
import numpy as np

df["profit_status"] = np.where(
    df["profit"] < -0.01,
    "Loss",
    np.where(df["profit"] <= 0.01, "Break-even", "Profit")
)

print(df["profit_status"].value_counts())
```

```python
discount_analysis = df.groupby("discountapplied").agg(
    orders=("ordernumber", "count"),
    total_sales=("sales", "sum"),
    total_profit=("profit", "sum"),
    avg_profit=("profit", "mean"),
    avg_margin=("profit_margin", "mean")
).reset_index()

discount_analysis
```

```python
discount_analysis["profit_margin_pct"] = (
    discount_analysis["total_profit"]
    / discount_analysis["total_sales"]
) * 100

discount_analysis
```

```python
channel_analysis = df.groupby("saleschannel").agg(
    orders=("ordernumber", "count"),
    total_sales=("sales", "sum"),
    total_cost=("total_cost", "sum"),
    total_profit=("profit", "sum"),
    avg_profit=("profit", "mean"),
    avg_delivery_days=("delivery_days", "mean")
).reset_index()

channel_analysis
```

```python
channel_analysis["profit_margin_pct"] = (
    channel_analysis["total_profit"]
    / channel_analysis["total_sales"]
) * 100

channel_analysis.sort_values(
    "total_profit",
    ascending=False
)
```

```python
warehouse_analysis = df.groupby("warehousecode").agg(
    orders=("ordernumber", "count"),
    total_sales=("sales", "sum"),
    total_profit=("profit", "sum"),
    avg_profit=("profit", "mean"),
    avg_delivery_days=("delivery_days", "mean"),
    avg_shipping_days=("shipping_days", "mean")
).reset_index()

warehouse_analysis["profit_margin_pct"] = (
    warehouse_analysis["total_profit"]
    / warehouse_analysis["total_sales"]
) * 100

warehouse_analysis.sort_values(
    "total_profit",
    ascending=False
)
```

```python
import numpy as np

df['profit_status'] = np.select(
    [
        df['profit'] > 0.01,
        df['profit'] < -0.01
    ],
    [
        'Profit',
        'Loss'
    ],
    default='Break-even'
)
```

```python
product_analysis = df.groupby('productid').agg(
    orders=('ordernumber', 'count'),
    quantity_sold=('orderquantity', 'sum'),
    total_sales=('sales', 'sum'),
    total_cost=('total_cost', 'sum'),
    total_profit=('profit', 'sum'),
    avg_discount=('discountapplied', 'mean'),
    avg_margin=('profit_margin', 'mean')
).reset_index()

product_analysis['profit_margin_pct'] = (
    product_analysis['total_profit'] /
    product_analysis['total_sales'] * 100
)

product_analysis.head()
```

```python
top_products = product_analysis.sort_values(
    'total_profit',
    ascending=False
).head(10)

top_products
```

```python
bottom_products = product_analysis.sort_values(
    'total_profit',
    ascending=True
).head(10)

bottom_products
```

```python
product_analysis.sort_values(
    ['total_sales', 'total_profit'],
    ascending=[False, True]
).head(20)
```

```python
high_sales_low_profit = product_analysis[
    (product_analysis['total_sales'] > product_analysis['total_sales'].median()) &
    (product_analysis['profit_margin_pct'] < product_analysis['profit_margin_pct'].median())
]

high_sales_low_profit
```
